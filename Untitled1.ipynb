```kotlin
// 1. Buat Dataset Sederhana (Data Penjualan)
val dfPenjualan = dataFrameOf(
    "Bulan" to listOf("Jan", "Feb", "Mar", "Apr", "Mei", "Jun"),
    "Penjualan" to listOf(150, 200, 180, 220, 300, 280)
)

// 2. Buat Grafik Batang (Bar Chart) pakai Kandy
dfPenjualan.plot {
    bars {
        x(Bulan)
        y(Penjualan)
    }
}
```
